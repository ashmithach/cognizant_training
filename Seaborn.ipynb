```python
import matplotlib.pyplot as plt
import numpy as np
products=["shampoo","paste","soap","brush","towel"]
year_2024=[100,300,400,200,150]
year_2025=[600,400,500,100,300]
year_2026=[300,220,480,440,200]
x=np.arange(len(products))
width=0.35
plt.title('2024vs 2025vs 2026')
plt.bar(x-width/2,year_2024,label='2024')
plt.bar(x-width/2,year_2025,label='2025')
plt.bar(x-width/2,year_2026,label='2026')
plt.xlabel("products")
plt.legend()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

months=['jan','feb','mar','apr','may']
sales=[100,700,500,400,300]
x=np.arange(len(products))
sns.lineplot(x=months,y=sales)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales trends")
plt.show()
```

```python
import plotly.express as px
import pandas as pd
df=pd.DataFrame({
    'months':['jan','feb','mar','apr','may'],
    'sales':[100,700,500,400,300]})
fig=px.line(df,x='months',y='sales',markers=True)
fig.show()
```

```python
df=pd.DataFrame({
    'semester':['sem1','sem2','sem3','sem4'],
    'marks':[1000,850,899,700]})
fig=px.line(df,x='semester',y='marks',markers=True)
fig.show()
```

```python
df=pd.DataFrame({
    'semester':['sem1','sem2','sem3','sem4'],
    'marks':[1000,850,899,700]})
fig=px.line(df,x='semester',y='marks',markers=False)
fig.show()
```
