```python
import numpy as np
num1=([7,8,9,10])
num2=([1,2,3,4])
num3=np.concatenate((num1,num2))
print(num3)
```

```python
arr=np.array([2,4,6,8])
arr1=np.where(arr==6)
print(arr1)
```

```python
arr=np.array([2,4,6,8])
arr1=np.sort(arr)
print(arr1)
```

```python
import numpy as np
marks=np.array([89,56,45,34,32])
print(marks[marks>=45])
```

```python
marks=np.sum([30,60,66,99,100])
print(marks)
```

```python
marks=np.average([77,45,67,88,90])
print(marks)
```

```python
marks1=np.min([23,45,56,67,89])
marks2=np.max([34,56,78,89,90])
print(marks1)
print(marks2)

```
