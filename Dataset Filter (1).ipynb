```python
datajoin=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["sanjana","pranavi","anjali"]})
datajoin3=pd.DataFrame({
    "ID":[1,2,3],
    "Marks":[67,32,54]})
result=pd.merge(datajoin,datajoin3, on="ID")
print(result)
```

```python
pd.merge(datajoin,datajoin3,on="ID",how="inner")
```

```python
leftjoin=pd.merge(datajoin,datajoin3, on="ID",how="right")
print(leftjoin)
```
