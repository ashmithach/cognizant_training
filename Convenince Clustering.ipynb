```python
customers=list(range(1,80))
sample=customers[::7]
print("Convenience sample",sample)
```

```python
employee={
          "Dany":67,
          "Drogo":69,
          "Jon":76
}
sample=[name for name,exp in employee.items() if(exp>=10)]
print("Judgement sample:",sample)
         
```

```python
networks={
        "Dany":["Drogo","Jon"],
        "Drogo":["Danerys"],
        "Jon":["Yigrette"]
}
sample=["Dany"]
sample.extend(networks["Dany"])
sample.extend(networks["Drogo"])
sample.extend(networks["Jon"])
print("snowball sampling:",sample)
              
```

```python
import pandas as pd
data={
      "customers_ID":range(1,101),
      "city":["essos"]*50+["pentos"]*50
}
result=pd.DataFrame(data)
e=result.sample(n=10)
print(e)
```
