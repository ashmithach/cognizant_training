```python
import re
email="dhruvithatyrell@gmail.com"
pattern=r"^[a-zA-Z0-9.!#$%&'+/=?^_`{|}~-]+@[a-zA-Z0-9-]+(?:\.[a-zA-Z0-9-]+)$"
if re.match(pattern,email):
    print("valid")
else:
    print("invalid")
```
